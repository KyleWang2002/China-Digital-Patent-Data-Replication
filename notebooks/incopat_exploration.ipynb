# IncoPat — Selenium interface exploration

This is a cleaned version of an earlier experiment navigating a dynamic patent-detail interface. It preserves the selector-based approach used in the original exploration; it is not an unattended scraper or a validated current-interface integration. An active, permitted IncoPat account is required for live use.

For a bounded CSV capture workflow, use `scripts/collect_incopat.py`. The saved five-row metadata excerpt is in `examples/incopat_ui_excerpt.csv`.

```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from bs4 import BeautifulSoup

driver = webdriver.Chrome()
driver.get("https://t.incopat.com/newLogin?locale=zh")
input("Log in manually, open an authorized patent-detail page, then press Enter: ")
wait = WebDriverWait(driver, 15)
```

```python
# Selectors adapted from the early IncoPat experiment; inspect current HTML if needed.
wait.until(EC.visibility_of_element_located((By.CSS_SELECTOR, "div.title span.overflow")))
wait.until(EC.visibility_of_element_located((By.CSS_SELECTOR, "div.fold_con")))

from patentrep.browser_capture import parse_detail_html
patent = parse_detail_html(driver.page_source)
{k: patent.get(k, "") for k in ("title", "application_no", "application_date", "applicant", "applicant_address")}
```
